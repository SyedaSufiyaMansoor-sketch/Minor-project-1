```python
# GROUP DNA - SIMPLE VERSION

import numpy as np

# Sample WhatsApp messages
messages = [
    ("Rahul", "bhai kya scene"),
    ("Rahul", "guys where are you"),
    ("Rahul", "come fast"),
    ("Priya", "Did everyone eat?"),
    ("Priya", "Please take care"),
    ("Aman", "I can't sleep"),
    ("Aman", "Anyone awake?"),
    ("Aman", "haha"),
    ("Karan", "Today I went to college and then I met my friends"),
    ("Karan", "After that we went to the canteen"),
    ("Neha", "GUYS LISTEN!!!"),
    ("Neha", "OMG THIS IS CRAZY!!!"),
    ("Vikas", "okay")
]


# --------------------------------
# 1. GROUP MEMBERS
# --------------------------------

members = []

for name, message in messages:
    if name not in members:
        members.append(name)

print("=" * 50)
print("        GROUP DNA REPORT")
print("=" * 50)

print("\nMembers:", len(members))

for name in members:
    print("-", name)


# --------------------------------
# 2. MESSAGE COUNT
# --------------------------------

count = {}

for name in members:
    count[name] = 0

for name, message in messages:
    count[name] = count[name] + 1


print("\nMESSAGE COUNT")
print("-" * 30)

for name in members:
    print(name, ":", count[name])


# --------------------------------
# 3. MOST ACTIVE PERSON
# --------------------------------

most_active = max(count, key=count.get)

print("\nMost Active Person:", most_active)


# --------------------------------
# 4. LEAST ACTIVE PERSON
# --------------------------------

least_active = min(count, key=count.get)

print("Least Active Person:", least_active)


# --------------------------------
# 5. TOP WORDS
# --------------------------------

words = {}

for name, message in messages:

    message = message.lower()

    for word in message.split():

        if word not in words:
            words[word] = 1
        else:
            words[word] = words[word] + 1


top_word = max(words, key=words.get)

print("\nMOST USED WORD")
print("-" * 30)

print(top_word, ":", words[top_word], "times")


# --------------------------------
# 6. NUMPY ACTIVITY
# --------------------------------

activity = np.array([
    count["Rahul"],
    count["Priya"],
    count["Aman"],
    count["Karan"],
    count["Neha"],
    count["Vikas"]
])

print("\nACTIVITY DATA")
print("-" * 30)

print(activity)

print("Total messages:", np.sum(activity))
print("Average messages:", np.mean(activity))


# --------------------------------
# 7. PERSONALITY DETECTION
# --------------------------------

print("\nPERSONALITY ARCHETYPES")
print("-" * 30)


for name in members:

    person_messages = []

    for sender, message in messages:

        if sender == name:
            person_messages.append(message)


    # Spammer
    if count[name] >= 3:
        personality = "THE SPAMMER"


    # Group Mom
    elif name == "Priya":
        personality = "THE GROUP MOM"


    # Night Owl
    elif name == "Aman":
        personality = "THE NIGHT OWL"


    # Storyteller
    elif name == "Karan":
        personality = "THE STORYTELLER"


    # Drama Queen
    elif name == "Neha":
        personality = "THE DRAMA QUEEN"


    # Ghost
    elif count[name] <= 1:
        personality = "THE GHOST"


    else:
        personality = "THE QUESTION MASTER"


    print(name, "→", personality)


# --------------------------------
# 8. FINAL RESULT
# --------------------------------

print("\n" + "=" * 50)
print("           GROUP DNA COMPLETE")
print("=" * 50)
```
